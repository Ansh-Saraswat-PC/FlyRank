## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset reveals a substantial opportunity for prioritization. While a large portion of the dataset consists of older, "stale" pages, only a specific subset of these pages actually have high baseline impressions and are actively trending downward. This gap proves why a targeted, multi-signal ranking system is necessary over simple age-based rules.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

total_pages = len(df)
declining_high_traffic = len(df[(df['trend_direction'] == 'down') & (df['impressions_90d'] >= 500)])
pct_declining = (declining_high_traffic / total_pages) * 100

stale_pages = len(df[df['content_age_days'] >= 180])
pct_stale = (stale_pages / total_pages) * 100

print(f"Total pages: {total_pages:,}")
print(f"High-traffic declining pages: {declining_high_traffic:,} ({pct_declining:.1f}%)")
print(f"Pages older than 180 days: {stale_pages:,} ({pct_stale:.1f}%)")
```
