## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (e.g., "update pages older than 180 days with >500 impressions") treats all pages that pass the threshold equally. If 300 pages pass the rule but the team can only update 50, the rule cannot tell you which 50 to pick. ML weighs multiple continuous signals simultaneously—balancing exact age, exact click-through-rate deficits, and exact impression volume—to output a granular score. This transforms an unmanageable binary bucket of "needs review" into a strictly ranked, capacity-aware priority list.
